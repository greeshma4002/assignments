```python
 import numpy as np
 import pandas as pd
 np.random.seed(42)
 n = 200
 df = pd.DataFrame({
     "order_id": range(1001, 1001 + n),
     "order_date": pd.date_range("2024-01-01", periods=n, freq="D"),
     "customer_name": np.random.choice(
          ["Anita", "Ravi", "Meera", "Arjun", "Sneha", "Karthik", "Divya",
           "Suresh"], n),
      "city": np.random.choice(
           ["Kochi", "Chennai", "Bengaluru", "Hyderabad", "Mumbai"], n),
      "category": np.random.choice(
          ["Electronics", "Clothing", "Grocery", "Books", "Furniture"], n),
      "quantity": np.random.randint(1, 10, n),
      "unit_price": np.random.randint(100, 5000, n),
     "payment_mode": np.random.choice(
          ["UPI", "Card", "Cash", "NetBanking"], n),
     })
 df.to_csv("sales_data.csv", index=False)
 print("File created with", len(df), "rows")
```

```python
def calculate_total(quantity, unit_price, discount_percent=0):
  subtotal = quantity * unit_price
  discount_amount = subtotal * (discount_percent / 100)
  final_amount = subtotal - discount_amount
  return final_amount
print(calculate_total(3, 500))
print(calculate_total(2, 1000, 10))
```

```python
import sqlite3
def load(df, db_name, csv_name):
  conn = sqlite3.connect(db_name)
  df.to_sql("clean_orders", conn, if_exists="replace", index=False)
  conn.close()
  df.to_csv(csv_name, index=False)
  print(f"Saved to SQLite table 'clean_orders' in {db_name} and to {csv_name}")
```

```python
import pandas as pd
def transform(df):
    df = df.copy()
    df = df.dropna()
    df["order_date"] = pd.to_datetime(df["order_date"])
    df["city"] = df["city"].str.strip().str.title()
    df["category"] = df["category"].str.strip().str.title()
    df["total_amount"] = df["quantity"] * df["unit_price"]
    df["month"] = df["order_date"].dt.month_name()
    return df
```

```python
raw_df = extract("sales_data.csv")
clean_df = transform(raw_df)
load(clean_df, "retail_etl.db", "clean_orders.csv")
```

```python
import pandas as pd
def transform(df):
    df = df.copy()
    df = df.dropna()
    df["order_date"] = pd.to_datetime(df["order_date"])
    df["city"] = df["city"].str.strip().str.title()
    df["category"] = df["category"].str.strip().str.title()
    df["total_amount"] = df["quantity"] * df["unit_price"]
    df["month"] = df["order_date"].dt.month_name()
    df["order_size"] = pd.cut(
        df["total_amount"],
        bins=[0, 10000, 25000, float("inf")],
        labels=["Small", "Medium", "Large"]
    ).astype(str)
    return df
```
